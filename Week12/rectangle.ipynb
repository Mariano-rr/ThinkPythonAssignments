```python
# IMPORT STATEMENTS
#(NONE)

# FUNCTION DEFINITIONS
#Create a function callled calculate_rectangle with 2 parameters: length, width
def calculate_rectangle(length, width):
# The function will calculate area by multiplying length and width
  area = length * width
# The function will calculate peri5meter by adding length and width and then multiplying it by 2
  perimeter = 2 * (length + width)
# The function will then return area, perimeter
  return (area, perimeter)

# PROGRAM LOGIC
#Ask the user to input the length of the rectangle and turn it into a float
length = float(input( 'Enter length: '))

#Ask the user to input the width of the rectangle and turn it into a float
width = float(input( 'Enter width: '))

#set the variables my_area, my_perimeter = calculate_rectangle function
my_area, my_perimeter = calculate_rectangle(length, width)

#print the my_area variable
#print the my_perimeter
print()
print( '--Result--' )
print( f'The area is: {my_area}' )
print( f'The perimeter is: {my_perimeter}' )
```
