```python
#IMPORT STATEMENTS
#(NONE)

#FUNCTION DEFINITIONS
#Create a function called filter with 2 parameters: Diction, option.
def Filter(Diction, option):
# The function will have a while loop with a try and except to catch errors
  while True:
    try:
#     It will use a for loop to print out every key contained in the dictionary
      for key in Diction.keys():
        print(f'Option Available: {key}')
      print()
#     Then, based on the option the user chose, it will sort by category or vendor and ask for user input on which specific category or vendor
      if option == 'c':
        userinput = input( f'Enter a Category: ').lower()
      else:
        userinput = input( f'Enter a vendor: ').lower()
#     The function will use the input given to find the values related to it and sum them all up before printing the total
      Total = sum(Diction[userinput])
      print(f'\nYour total for {userinput} was ${Total:.2f}\n')
      break
#   if an error occurs, it will print an error message
    except KeyError:
      print( f'Choice not found, try again.\n' )

#Create a function called AmountSecurity with no parameters
def AmountSecurity():
# The function will have a while loop along with a try and except to catch incorrect inputs
  while True:
    try:
#     The function will ask for the amount, and if no errors occur, it will return the value
      Amount = float(input( 'Enter amount: '))
      return Amount
#   if an error occurs, it will print an error message
    except ValueError:
      print( 'Incorrect input.\n' )

#PROGRAM LOGIC
#Create a Category Dictionary and a Vendor Dictionary
CatDictionary = {}
VenDictionary = {}

#Create a while loop and print the main user interface and ask the user to select a choice
while True:
  print( '--Expense Tracker-- ')
  print( '1. Add new expense' )
  print( '2. Get Totals ')
  print( '3. Exit\n' )
#The program will check if there are things inside the dictionary and if not, will prompt user to pick 1 or 3.
  if not CatDictionary:
    print( f'Nothing to Filter. Pick options 1 or 3.\n' )
    choice = int(input( 'Enter a number: '))
    if choice == 2:
      choice = 1
    elif choice == 3:
      break
    else:
      choice = 1
  else:
    choice = int(input( 'Enter a number: '))
# If the choice was 1, it will prompt the user for a Category, Vendor, Amount (by calling the function AmountSecurity), and date.
  if choice == 1:
    Category = input( 'Enter a category: ').lower()
    Vendor = input( 'Enter a vendor: ').lower()
    Amount = AmountSecurity()
    Date = input( 'Enter date (2024-03-31)')
    print()
#   the program will then check to see if the vendor or category given by the user already exists within a dictionary
    if Category in CatDictionary:
#     If they don’t it will create a list, add the amount to it, and make it the value of the dictionary, if it does, it will add the amount to the existing list.
      CatDictionary[Category].append(Amount)
    else:
      CatDictionary[Category] = [Amount]
    if Vendor in VenDictionary:
      VenDictionary[Vendor].append(Amount)
    else:
      VenDictionary[Vendor] = [Amount]

# If the user chose option 2, it will print another user interface for sorting and ask for user input.
  elif choice == 2:
    print()
    print( 'Totaling options')
    print( 'A. All Expenses')
    print( 'C. Filter by Category')
    print( 'V. Filter by Vendor\n')
    option = input( f'Enter a letter: ')
    print()
#   If the user chose option a, it will take the values of a dictionary and total them up via a for loop before printing it
    if option == 'a'.lower():
      total = 0
      for values in CatDictionary.values():
        total += sum(values)
      print(f'Total Expenses: ${total:.2f}\n')
#   If the user chose the option c or v, it will call the function Filter
    elif option == 'c'.lower():
      Filter(CatDictionary, option)
    elif option == 'v'.lower():
      Filter(VenDictionary, option)
    else:
      print( 'Invalid option' )
# If the choice was 3, the program will exit and end.
  elif choice == 3:
    break
# If any other choice was entered, it will print invalid option.
  else:
    print( f'Invalid choice.\n' )
    continue
```
