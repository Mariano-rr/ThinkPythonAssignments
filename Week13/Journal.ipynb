```python
#IMPORT STATEMENTS
#(NONE)

#FUNCTION DEFINITIONS
#Create a function named write_entry with two parameters: filename, entry
def write_entry(filename, entry):
# The function will open the file in append mode
  with open(filename, 'a') as file:
#   it will then write an entry with a \n at the end to write it in a new line.
    file.write(f'{entry}\n')

#Create a function named read_journal with a single parameter: filename
def read_journal(filename):
# The function will try to open the file in read mode
  try:
    with open(filename, 'r') as contents:
#     It will then print the contents
      print( f'{contents.read()}')
# if it encounters the file not found error, it will print an error message and not do anything
  except FileNotFoundError:
    print( 'Error! No entries written.' )

#PROGRAM LOGIC
#Set filename = to Journal.txt
filename = 'journal.txt'

#Create a while true loop that asks the user if they want to write in or read their journal
while True:
  key = input( '(W)rite, (R)ead, or (Q)uit? ' ).lower()
# If the user types r, it will print 'your journal' and the read_journal function is called
  if key == 'r':
    print( f'-- Your Journal --')
    read_journal(filename)
# If the user types w, the user will be prompted to write an entry and the program will display a message afterwards
  elif key == 'w':
    thoughts = input( 'Enter your entry: ')
    print( 'Entry saved!' )
    print()
#   Then the write_entry function will be called using the entry.
    write_entry(filename, thoughts)
# If the user types q, the program ends
  elif key == 'q':
    break
  else:
    continue
```
