```python
#IMPORT STATEMENTS
#(NONE)

#CLASS DEFINITIONS
#Create a class called Book
class Book():
# Create an __init__ method that has the attributes name and author
  def __init__(self, title, author):
    self.title = title
    self.author = author
    #Add a third attribute that is called is_available and is set to True
    self.is_available = True
# Create a __str__ method with (self)
  def __str__(self):
#   If is_available is true, the status is available, if false, status is checked out
    if self.is_available == True:
      status = 'Available'
    else:
      status = 'Checked out'
#   Return a string that shows title, author, and status
    return (f'{self.title} by {self.author} - {status}')
# Create a checkout method with (self)
  def checkout(self):
#   It will set is_available to false and return a string saying checking out book
    self.is_available = False
    return (f'Checking out {self.title}...')
# Create a return_book method with (self)
  def return_book(self):
    #It sets is_available to true and returns a string saying returning book
    self.is_available = True
    return (f'Returning {self.title}...')

#Create a my_book object
my_book = Book( 'Think Python', 'Allen Downey' )
#Create a my_other_book object
my_other_book = Book( 'The Hobbit', 'J.R.R. Tolkien' )

#print the books
print( '-- Library Status --' )
print(my_book)
print(my_other_book)
print()
#use the checkout method
print(my_other_book.checkout())
print()
#print the books
print( '-- Updated Status --' )
print(my_book)
print(my_other_book)
```
