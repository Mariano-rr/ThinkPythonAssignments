```python
from dataclasses import dataclass

# Permanent file name for completed tasks
HISTORY_FILE = "task_history.txt"

@dataclass
class Task:
    """Structure to hold task data."""
    name: str
    priority: int

    @property
    def priority_label(self) -> str:
        """Map priority integers to readable labels."""
        labels = {1: "High", 2: "Medium", 3: "Low"}
        return labels.get(self.priority, "Unknown")

class TaskManager:
    """Handles the storage, completion, and display of tasks."""
    def __init__(self):
        self.tasks: list[Task] = []

    def add_task(self, name: str, priority: int) -> None:
        """Adds a new task and sorts the list by priority."""
        self.tasks.append(Task(name=name, priority=priority))
        self.tasks.sort(key=lambda t: t.priority)
        print(f"\n[✓] Task '{name}' added successfully!\n")

    def complete_task(self, task_index: int) -> None:
        """Removes a task from the list and logs it to a file."""
        if 0 <= task_index < len(self.tasks):
            # Remove the task from the active list
            completed_task = self.tasks.pop(task_index)

            # Save the completed task to the history file
            with open(HISTORY_FILE, "a", encoding="utf-8") as file:
                file.write(f"{completed_task.name}\n")

            print(f"\n[✓] Task '{completed_task.name}' completed and saved to history!\n")
        else:
            print("\n[X] Invalid task number.\n")

    def display_tasks(self) -> None:
        """Prints the tasks in a formatted grid."""
        if not self.tasks:
            print("\n[!] No tasks available yet.\n")
            return

        print("\n" + "=" * 45)
        print(f"{'ID':<5} {'Task Name':<25} {'Priority':<10}")
        print("=" * 45)

        for idx, task in enumerate(self.tasks, start=1):
            print(f"{idx:<5} {task.name:<25} {task.priority_label:<10}")

        print("=" * 45 + "\n")

def main():
    manager = TaskManager()
    print("Welcome to your Personal Task Manager!")
    print("-" * 38)

    while True:
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. Exit")

        choice = input("\nChoose an option (1-4): ").strip()

        if choice == "1":
            name = input("\nEnter task name: ").strip()
            while not name:
                name = input("Task name cannot be empty. Enter task name: ").strip()

            priority = input("Enter priority (1=High, 2=Medium, 3=Low) [Default 2]: ").strip()
            if priority not in ["1", "2", "3"]:
                priority = "2"

            manager.add_task(name, int(priority))

        elif choice == "2":
            manager.display_tasks()

        elif choice == "3":
            manager.display_tasks()
            if manager.tasks:
                try:
                    task_num = int(input("Enter the ID of the task to complete: ").strip())
                    # Subtract 1 to match Python's zero-based indexing
                    manager.complete_task(task_num - 1)
                except ValueError:
                    print("\n[X] Please enter a valid number.\n")

        elif choice == "4":
            print("\nGoodbye! Exiting task manager.")
            break
        else:
            print("\n[X] Invalid option. Please enter 1, 2, 3, or 4.\n")

if __name__ == "__main__":
    main()
```
