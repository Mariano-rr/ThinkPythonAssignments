```python
# IMPORT STATEMENTS
#(NONE)


# FUNCTION DEFINITIONS
#Create a function called car_classification with a parameter: letter
#   The function will take the letter inputted and return the unabbreviated version for the final report

#Create another function called mileage with the parameter: mile
#   The function will check the mile variable and if it is less than 100, returns 0
#   If it it more than 100, it will remove 100 then multply the remaining by .35 and return the value

#Create a function called days_rented with 2 paramaters: days and letter
#   The function will take the letter you inputted and set the letter variable to the number that goes with it
#   Then it will multiply it by the days and return its value

#Create another function called final_bill with 3 paramters: mile, days, and letter
#   It will add the mileage and days_rented functions together and return their values.

# PROGRAM LOGIC
#Create a while loop and ask for the user to enter a valid letter
#The if statement checking the letter will return an error if the letter is not valid.
#
#Ask the user for the start and end odoemter values then subtract them and store it in the variable mile
#Ask the user for the days rented and store it in a variable
#
#Create an if statement to find the miles charged for the final report
#If it is less than 100, miles_charged becomes 0 else subtract 100 from miles and store it in the miles_charged variable
#
#Print the final report as follows:
#print classification: and the car_classification function
#print total miles: and str mile variable
#print charged miles: and str miles_charged variable
#print the total bill: and str final_bill function with its parameters.

while True:
  letter = input( 'Enter classification code (B, S, L): ' )
  if letter == 'B':
    break
  elif letter == 'S':
    break
  elif letter == 'L':
    break
  else:
    print( 'Incorrect input, please try again.' )

def car_classification(letter):
  if letter == 'B':
    return 'Budget'
  elif letter == 'S':
    return 'Standard'
  else:
    return 'Luxury'

start = int(input( 'Start Odometer: ' ))
end = int(input( 'End Odometer: ' ))
mile = end - start

def mileage(mile):
  if mile <= 100:
    return 0
  else:
    mile = ((mile - 100) * .35)
    return mile

days = int(input( 'Days Rented: ' ))

def days_rented(days, letter):
  if letter == 'B':
    letter = 40
  elif letter == 'S':
    letter = 60
  else:
    letter = 80
  return days * letter

def final_bill(mile, days, letter):
   return mileage(mile) + days_rented(days, letter)

if mile <= 100:
  miles_charged = 0
else:
  miles_charged = mile - 100

print()
print( '--- RENTAL SUMMARY ---')
print( 'Classification: ' + car_classification(letter))
print( 'Total Miles: ' + str(mile))
print( 'Charged Miles: ' + str(miles_charged))
print( 'Total Bill: $' + str(final_bill(mile, days, letter)))
```
